### to handle missing values in dataset of file

```python
# same above function
def convert_to_float (line): 
    values=[]
    for individual_val in line.strip().split(','):     # strip remove new line char & split -> splits string into list of words sep by comma 
        if individual_val=='': 
            values.append(0.0)                                    #then list is traversed
        else:
            values.append(float(individual_val))
    return values   
```

## **combining everything from reading to handling missing values to making dictionary of dataset

```python
# function to parse header

def parse_header(header):
    return header.strip().split(',')                              # list of token of individual header names


# function to parse values by handling missing values

def parse_values(data_line):
    values = []                                                    # niche wale ke baad hum nyi list de rhe after handling missing val (handled)
    for val in data_line.strip().split(','):                       # milega -> list of  of token of values from data_line (include missing val)
        if val == '':
            values.append(0.0)
        else:
            values.append(float(val))
    return values


# function to create dict

def create_dict(header,values):
    d={}
    for head,value in zip(header,values):
        d[head]=value
    return d


# function to read and accsess data/content of file

file_name = input('whats the file name?')
def read_csv(file_name):
    final_result=[]
    with open (file_name,'r') as f:
        
        f_content=f.readlines()                                        # list os strings of data lines

        # worked at header
        header = parse_header(f_content[0] )                           #  header ke function ko header ke liye call kr idya

        # worked at values
        for all_data_lines_oneBYone in f_content[1:]:                   # 1 i.e headear chodke at idx 0 humne values ko liye from idx 1
            values = parse_values(all_data_lines_oneBYone)                 # fir ek ke data line ke liye pare values function ka call kiya

            # creating dictionary
            dictionary_of_content=create_dict(header,values)

            # add dictionary to final result
            final_result.append(dictionary_of_content)
    return final_result


print(read_csv(file_name))
   
    
```
