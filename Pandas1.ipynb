## Analyzing data from dataframes


```python
## let's answer some questions about our data
```

```python
import pandas as pd

dataset_df = pd.read_csv ("italy-covid-daywise .csv")
dataset_df
```

```python
# Q1. What is the total number of reported cases and deaths related to covid-19 in italy ?
```

```python

what_col_we_have = dataset_df.columns
what_col_we_have
```

```python
reported_cases = dataset_df.new_cases.sum()
reported_cases
```

```python
reported_deaths = dataset_df.new_deaths.sum()                 #  -> .sum()  ignores NaN
reported_deaths
```

```python
print("the number of reported cases are {} and of reported deaths are {}".format(reported_cases,reported_deaths.))
```

```python
# Q2. What is the overall death ratio ( ratio of reported deaths to reported cases )
```

```python
death_rate = dataset_df.new_deaths.sum() / dataset_df.new_cases.sum()
death_rate
```

```python
print( "The overall death ratio is {:.2f} % .".format(death_rate))         # to format decimal we use {:.2f} in print statement
```

```python
# Q3. What is the overall number of tests conducted? A total of 935310 tests were conducted before daily test numbers were reported.
```

```python
# ques ka mtlb itne no. of test conduct ho ja rhe the report hone se pehel hi
# therefore overall no. of test hua = jitne conduct hue + jitne report hue
```

```python
no_of_test_conducted_before_reporting = 935310
no_of_tests_reported = dataset_df.new_tests.sum()
no_of_tests_reported
```

```python
total_overall_tests_conducted = no_of_test_conducted_before_reporting + no_of_tests_reported
total_overall_tests_conducted
```

```python
print("Total overall tests condcuted are {}.".format(total_overall_tests_conducted))
```

```python
# Q3. What fraction of tests resturned a positive result
```

```python
what_col_we_have = dataset_df.columns
what_col_we_have
```

```python
positive_results = reported_cases/ total_overall_tests_conducted
positive_results
```

```python
print("The total no. of positive results are {:.2f}.".format(positive_results))
```

## Grouping and Aggregation

```python
# seeing the data monthly wise

# s1. making monthly grpoups

dataset_by_months_grp = dataset_df.groupby("month")                 # 1st - grouping
```

```python
# s2. aggregation

dataset_by_months_grp[["new_cases","new_deaths","new_tests"]].sum()       # 2nd - column selection  - same flow as in sql
```

```python
# what were the average no. of cases , deaths, tests weekly

weekly_grp = dataset_df.groupby("week")

weekly_grp[["new_cases","new_deaths","new_tests"]].mean()
```
