### without numpy
 given some climate data for a region, we can now predict what the yield of apples in the region might look like. Here's some sample data.
region  |  temp (F) |   rainfall(mm)  |   humidity(%)
kanto   |    73     |        67       |        43
johto   |    91     |        88       |        64
hoenn   |    87     |        134      |        58
sinnoh  |    102    |        43       |        37
unova   |    69     |        96       |        70





```python


kanto_region=[73 , 67 , 43]
johto_region=[91 , 88 , 43]
hoenn=[87 , 134 , 58]
sinnoh=[102 , 43 , 37]
unova=[69 , 96 , 70]
wts=[0.3,0.2,0.5]

# for x in kanto_region:
    #print(x)

#let's make general function to find out apple yield based on these factors
def apple_yield(region,wts):
    yield_calc=0
    for x,w in zip(region,wts):
        yield_calc=yield_calc+(x*w)
    return yield_calc
apple_yield(kanto_region,wts)
```

### let's compare time taken using numpy and without

```python
#python list
arr1_list = list(range(1000000))
arr2_list = list(range(1000000,2000000))

#numpy arrays
import numpy as np
arr1_np = np.array(arr1_list)
arr2_np = np.array(arr2_list)
```

```python
# using list
import time
start_time=time.perf_counter()
result=0
for x,y in zip(arr1_list,arr2_list):
    result = result + (x*y)
result
end_time=time.perf_counter()
print("time taken :",end_time - start_time)
```

```python
# using numpy
import time
start=time.perf_counter()
np.dot(arr1_np,arr2_np)
end=time.perf_counter()
print("time taken :",end-start)
```

```python
import numpy as np
climate_data = np.genfromtxt("Bank_Personal_Loan_dataset.csv",delimiter=',',skip_header=1)
```

```python
climate_data.dtype
```

```python
climate_data
```

```python
climate_data.shape
```

```python
# now let's predict yield of apple , considering weights
weights = np.array([0.3,0.2,0.5])
```

```python
weights.shape
```

```python
yield_calc = climate_data @ weights        # matrix multiplication
```

```python
yield_calc
```

```python
yield_calc.shape
```

```python
# we can now add yields(result) back to climate_data(file) as a fourth column using np.concatenate function
# np.concatenate(dataset, yields , where to add row/col -> axis = 1 for col and 0 for row
climate_result = np.concatenate((climate_data, yield_calc.reshape(8766,1)) , axis = 1)
climate_result
```

```python
# WRITING/SAVING RESSULT IN THE FILE
np.savetxt("climate_result.csv",climate_result, delimiter="," ,fmt="%.2f" , header ="Temp_Max,Temp_Min,Windspeed_Max,Yield_apples")
```

# NOTES

```python
# Arithmetic op -> +, -, *, /, //, % 
# Broadcasting -> diiferent dimension but same shape -> artihmetic valid and can be applied 1d ko internally 2d bna ke operation krta hai
# 2D shape (row,col)    and  1D shape (ele,) i.e no row no col
# 1D - (no. of ele,)
# 2D - (no. of row,no. of col)
# 3D - (no. of layer,no. of row,no. of col)
# a=np.array([1,2,3,]) or a=np.array([1,2,3])   -> no diffenrence in shape comma is optional makes no effect
# np.array([[1,2,3,4],
#           [5,6,7]])      inhomogenous shape error ( bcz row length is diffferent )
# Comparison op -> == , != , < , > , <= , >=
# in comparison -> dtype is bool and matrix formed is matrix of boolean values
# (arr1 == arr2).sum()  -> to coun no. of equal elements in 2 arrays
# False -> 0
# True -> 1
# INDEXING
#                        col0  col1  col2   col3
# arr3d = np.array ( [ [[ 1  ,  2  ,  3  ,  4],                row0    |>
#                       [5   ,  6  ,  7  ,  8  ]],               row1    |> layer 1
#
#                       [[9  ,  10  ,  11  ,  12],            row0    |> 
#                       [13  ,  14  ,  15  ,  16]],           row1    |> layer 2
#
#                       [[17  ,  18  ,  19  ,  20],           row0    |> layer 3
#                       [21  ,  22  ,  23  ,  24]] ] )        row1    |>
#(slicing,slicing,slicing) -> no dimension reduction
#(slicning,ele,_)  -> dimension reduction -> no. of ele = no of dimensions reduced
# (:) -> return all
# when you give extra index than in array -> index error
```
