# Let's try to understand sales trend what happened it increased or decreased

## Step-1 > finding sales column

```python
dataset1.columns
```

## shifting

```python
dataset1["shift_by_1"]=dataset1["Sales"].shift(1)
```

```python
dataset1
```

### now afrer shift we got nan in shift column - so let's fill 0

```python
dataset1["shift_by_1"].isnull()
```

```python
dataset1["shift_by_1"].fillna(0,inplace=True)
```

## calculating change (kl se aaj tk me kitna chnage hua)

```python
dataset1["sales_change"]=dataset1["Sales"]-dataset1["shift_by_1"]
```

```python
dataset1
```

## now for better understanding marking it with up and down

```python
dataset1["sales_up_or_down"]=dataset1["sales_change"].apply(lambda x:"up" if x>0 else "down")
```

```python
dataset1
```

```python
dataset1[["Sales","sales_change","sales_up_or_down"]]
```
