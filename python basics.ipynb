```python
# combining both to get total interest
import math
loan_amt=100000
loan_duration=10*12
interest_rate=0.09/12

# ek baar with 9% interest nikal le and ek baar with 0% interest -> jitna differnce hoga unta extra jayega fro interest other than amt
def emi_calc(p,r,n):
    try:
        emi_with_interest=math.ceil((p*r*((1+r)**n)) / (((1+r)**n)-1))
        return emi_with_interest
    except ZeroDivisionError:
        emi_without_interest=loan_amt/loan_duration
        return emi_without_interest
```

```python
emi_with_interest=math.ceil(emi_calc(p=loan_amt,r=interest_rate,n=loan_duration))
emi_with_interest
```

```python
emi_without_interest=math.ceil(emi_calc(p=loan_amt,r=0,n=0./12))
emi_without_interest
```

```python
total_interest = (emi_with_interest - emi_without_interest)*10*12
print("the total_interest is $ {}".format(total_interest))
```
